```python

from functools import reduce
from functools import cmp_to_key

# =====================================
# CHALLENGE 1: Bank Transactions
# =====================================

transactions = [
    (-1200, 'debit'),
    (2500, 'credit'),
    (-100, 'debit'),
    (850, 'credit'),
    (-250, 'debit'),
    (1500, 'credit'),
    (-300, 'debit'),
    (5000, 'credit'),
    (-850, 'debit'),
    (1000, 'credit')
]

# Exercise 1: Filter credit and debit transactions
credits = list(filter(lambda x: x[1] == 'credit', transactions))
debits = list(filter(lambda x: x[1] == 'debit', transactions))

print("Credits:", credits)
print("Debits:", debits)


# Exercise 2: Sort debit transactions descending
sort_descending = lambda a, b: a[0] > b[0]

sorted_debits = sorted(
    debits,
    key=lambda x: x[0],
    reverse=True
)

print("Sorted debits:", sorted_debits)


# =====================================
# CHALLENGE 2: Interest Calculation
# =====================================

# Exercise 1
balances = [100, 50, -25, 1000, -10]
interest_rate = 0.05

new_balances = list(
    map(lambda balance: balance * (1 + interest_rate), balances)
)

print("Balances after interest:", new_balances)


# Exercise 2
accounts = [
    {'balance': 1000, 'interest_rate': 0.02},
    {'balance': 2000, 'interest_rate': 0.01},
    {'balance': 500, 'interest_rate': 0.03}
]

updated_accounts = list(
    map(
        lambda account: {
            'balance': account['balance'] * (1 + account['interest_rate']),
            'interest_rate': account['interest_rate']
        },
        accounts
    )
)

print("Updated accounts:", updated_accounts)


# =====================================
# CHALLENGE 3: Balance Reduction
# =====================================

# Exercise 1: Sum negative balances
negative_balances = list(
    filter(lambda balance: balance < 0, new_balances)
)

total_negative = reduce(
    lambda a, b: a + b,
    negative_balances,
    0
)

print("Total negative balances:", total_negative)


# Exercise 2: Calculate remaining balances
accounts = [
    {'balance': 1000, 'withdrawals': [100, 50, 200]},
    {'balance': 2000, 'withdrawals': [300, 100]},
    {'balance': 500, 'withdrawals': [50, 100, 75]}
]

def calculate_balance(account):
    return account['balance'] - sum(account['withdrawals'])

remaining_balances = list(
    map(calculate_balance, accounts)
)

print("Remaining balances:", remaining_balances)
```
