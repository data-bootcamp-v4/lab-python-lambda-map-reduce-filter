# Lab | Lambda Functions, Map, Reduce, Filter

Objective: The objective of this lab is to help students become familiar with using anonymous lambda functions and mapping, reducing, and filtering methods in Python.

Las funciones Lambda, mapear, reducir y filtrar están relacionadas con la programación funcional en Python. 

Las **funciones Lambda** son funciones anónimas en Python, lo que significa que no es necesario definirlas con un nombre. Por lo general, se usan para funciones breves de una línea que no se usarán en ninguna otra parte del código. 

Las funciones Lambda pueden tomar cualquier número de argumentos, pero solo pueden contener una única expresión. A menudo se usan en combinación con otras funciones integradas como mapear, reducir y filtrar para realizar operaciones en listas u otros iterables.

**Mapa** es una función que aplica una función determinada a cada elemento de una lista, devolviendo una nueva lista con los valores transformados. Es una forma útil de aplicar la misma operación a cada elemento de una lista, sin tener que escribir un bucle for.

**Reducir** es una función que aplica una función determinada a los dos primeros elementos de una lista, luego aplica la misma función al resultado y al siguiente elemento, y así sucesivamente, hasta que haya reducido la lista a un solo valor. Es útil para realizar operaciones acumulativas como sumar o multiplicar una lista de números.

**Filtro** es una función que toma una función y una lista, y devuelve una nueva lista que contiene solo los elementos de la lista original para los cuales la función devuelve Verdadero. Es una forma útil de extraer selectivamente elementos de una lista en función de una determinada condición.

Todos estos conceptos se pueden utilizar para hacer que su código sea más conciso y expresivo, especialmente cuando se trabaja con listas u otros iterables. También son útiles en tareas de manipulación de datos, como limpiar y transformar datos en un DataFrame de pandas. Al practicar estos conceptos en el laboratorio, comprenderá mejor cómo utilizarlos en una variedad de tareas de programación.

### Exercise 1

```python
debit = list(filter(lambda t: t[1] == 'debit', transactions))
credit = list(filter(lambda t: t[1] == 'credit', transactions))
print(credit)
```

## Challenge 2:  Interest Calculation

### Exercise 1

Write Python code to take a list of bank account balances, and returns a new list containing the balance after one year of interest has been added. Use the map function to apply this function to the list of bank accounts, and take an interest rate of 0.05.

```python
# create list of bank account balances
balances = [100, 50, -25, 1000, -10]
```

```python
task = 0.05

def agregar_interes(balances):
    return balances * (1+ task)

balances = [100, 50, -25, 1000, -10]
balances_con_task = list(map(agregar_interes, balances))
print(balances_con_task)
```

### Exercise 2

Write Python code to take a list of bank account dictionaries, each containing the account balance and interest rate, and returns a new list of dictionaries containing the balance after one year of interest has been added. Use the map function to apply this function to the list of bank accounts.

```python
accounts = [
    {'balance': 1000, 'interest_rate': 0.02},
    {'balance': 2000, 'interest_rate': 0.01},
    {'balance': 500, 'interest_rate': 0.03},
]
```

```python
def agregar_interes_cuenta(accounts):
    nuevo_balance = accounts['balance'] * (1 + accounts['interest_rate'])
    return {'balance' : nuevo_balance, 'interest_rate': accounts['interest_rate']}

cuentas_con_interes = list(map(agregar_interes_cuenta, accounts))
print(cuentas_con_interes)
```

## Challenge 3: Balance Reduction

### Exercise 1

Escriba código Python para tomar la nueva lista de saldos de cuentas bancarias (lista de saldos después de aplicar una tasa de interés de 0.05, resultado del Desafío 1 Ejercicio 1) e imprima la cantidad total de saldos negativos. Utilice la función de filtro y reducción.

```python
from functools import reduce
transacciones = [2500, 850, 1500, 5000, 1000]
transacciones_con_interes =list(map(lambda x:x*1.05, transacciones))
print(transacciones_con_interes)

total = reduce(lambda a,b : a+b, transacciones_con_interes)
print(total)
```

### Exercise 2

Write a Python function called calculate_balance that takes a bank account dictionary as an argument and returns the remaining balance after subtracting all the withdrawals.

Then, use the map function and the calculate_balance function to apply it to the list accounts. This should give you a list of remaining balances after all the withdrawals have been subtracted.

```python
accounts = [
    {'balance': 1000, 'withdrawals': [100, 50, 200]},
    {'balance': 2000, 'withdrawals': [300, 100]},
    {'balance': 500, 'withdrawals': [50, 100, 75]},
]

def calculate_balance(accounts):
    return accounts['balance'] - sum(accounts['withdrawals'])

remaining_balances = list(map(calculate_balance, accounts))
print(remaining_balances)
```
