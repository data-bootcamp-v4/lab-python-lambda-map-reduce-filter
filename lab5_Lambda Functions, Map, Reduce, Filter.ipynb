```python
transactions = [
    (-1200, 'debit'),
    (2500, 'credit'),
    (-100, 'debit'),
    (850, 'credit'),
    (-250, 'debit'),
    (1500, 'credit'),
    (-300, 'debit'),
    (5000, 'credit'),
    (-850, 'debit'),
    (1000, 'credit')
]

debits = list(filter(lambda transaction: transaction[1] == 'debit', transactions))

print(debits)
```

```python
sort_descending = lambda transaction: transaction[0]

sorted_debits = sorted(debits, key=sort_descending, reverse=True)

print(sorted_debits)
```

```python
accounts = [
    {'balance': 1000, 'interest_rate': 0.02},
    {'balance': 2000, 'interest_rate': 0.01},
    {'balance': 500, 'interest_rate': 0.03}
]

updated_accounts = list(map(
    lambda account: {
        'balance': account['balance'] * (1 + account['interest_rate']),
        'interest_rate': account['interest_rate']
    },
    accounts
))

print(updated_accounts)
```

```python
from functools import reduce

negative_balances = list(filter(lambda balance: balance < 0, new_balances))

total_negative = reduce(lambda x, y: x + y, negative_balances)

print(total_negative)
```

```python
accounts = [
    {'balance': 1000, 'withdrawals': [100, 50, 200]},
    {'balance': 2000, 'withdrawals': [300, 100]},
    {'balance': 500, 'withdrawals': [50, 100, 75]}
]

def calculate_balance(account):
    return account['balance'] - sum(account['withdrawals'])

remaining_balances = list(map(calculate_balance, accounts))

print(remaining_balances)
```
